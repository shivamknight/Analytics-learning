```python
import pandas as pd
import numpy as np
```

```python
# cresting series
ages = pd.Series([22,25,30],name="age")

# creating dataframe
data ={
    "Name":  ["Alice","Bob","Charlie","David"],
    "Age":   [25,30,np.nan,22],
    "Dept" : ["Hr","It","It","Hr"],
    "Salary":[50000,62000,70000,48000],
}
df =pd.DataFrame(data)

# ages
df
```

```python
print(df.head(1))
print(df.shape)
print(df.info)
print(df.dtypes)
```

```python
print(df)
print(df.describe())
```

```python
print(df)


df.loc[0:2,["Name","Salary"]]
df.iloc[0:2,0:2]
df.at[0,"Name"]
df.iat[0,1]
```

#### Boolean Filtering & Logical Operators

```python
# it high earners above 60k
it_high_earner = (df["Dept"]=="It") & (df["Salary"] >= 60000)
df[it_high_earner]

# filter:  not in hr 
not_hr = ~(df["Dept"] =="Hr")
# not_hr =  (df["Dept"] !="Hr")
df[not_hr]
```

```python
df["Dept"].nunique()
df["Dept"].value_counts()
df.isnull().sum()
df.isna().sum()
```
