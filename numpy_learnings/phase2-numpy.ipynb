## Array reshaping

```python
arr=np.arange(12)
print(f"original array: \n{arr}")  ##original array

reshaped=arr.reshape((3,4))
print(f"reshaped array: \n{reshaped}")  ##reshaped array

flat= arr.flatten()
print(f"Falettened aarray: \n{flat}")   ## faltening the array of any dimensions

raveled=arr.ravel()    ## it returns the original array 
print(f"Ravel return the original array: \n{raveled}")

# array transpose 
trans= reshaped.T
print(f"Transpose of reshaped: \n{trans}")
```

## sorting using Numpy

```python
## basic sorting
unnsort=np.array([3,4,2,4,1,5,7,1,8,6,9,2])
print(f"Sorted array: {np.sort(unnsort)}")

## sorting in 2d array  in column
arr_2d_unsorted=np.array([
    [6,7,9],
    [1,5,2],
    [8,3,4]
])
print(f"Unsorted 2d array: \n{arr_2d_unsorted}")
##sorting all rows
print(f"sorted 2d array(rows): \n{np.sort(arr_2d_unsorted,axis=1)}")
##sorting all colums
print(f"sorted 2d array(columns): \n{np.sort(arr_2d_unsorted,axis=0)}")
```

## Fancy indexing v/s np.where()

```python
numbers = np.array([5,4,10,2,7,8,1,9,3,6,0])

indices=[0,2,4]
print(f"NUMBERs with indices: {numbers[indices]}")

where_result=np.where(numbers>5)
print(f"{numbers[where_result]}")
```

```python
num=[1,3,5,7,9]
# num2=[2,4,6,8,10]
conditon_array=np.where(numbers>5,"True","False")
print(conditon_array)
conditon_array=np.where(numbers>5,numbers,numbers)
print(conditon_array)
```

```python
# 
original=np.array([[1,2],[3,4]])



new_row=np.array([[5,6]])
with_new_row=np.vstack((original,new_row))
# vstack is used to add vertically add new roq data
print(original)
print(with_new_row)

# hstack is used to add horizonnatally  new column data 
new_column=np.array([[7],[8],[9]])
with_new_column=np.hstack((with_new_row,new_column))
print(f"H : \n{with_new_column}")
```

```python
# Data structure: [restaurant_id,2021,2022,2023,2024]

sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hut
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])

print("===========zomato analysis============")

# total sales per year 
print(f"total sales per year:-")
print(f"2021: {sum(sales_data[:,1])}")
print(f"2022: {sum(sales_data[:,2])}")
print(f"2023: {sum(sales_data[:,3])}")
print(f"2024: {sum(sales_data[:,4])}")
# yearly total as in array:-
yearly_sale=np.sum(sales_data[:,1:],axis=0)
print(yearly_sale)
# minimum sale per restaurent:--
print("minimum sale per year:  ")
min_sale=np.min(sales_data[:,1:],axis=1)
print(min_sale.reshape(-1,1))
# maximum sale per restaurent:--
print("maximum sale per year:  ")
max_sale=np.max(sales_data[:,1:],axis=1)
print(max_sale.reshape(-1,1))
```

```python
cumsum=np.cumsum(sales_data[:,1:],axis=1)
print(cumsum)

plt.Figure(figsize=(10,6))
plt.plot(np.mean(cumsum,axis=0))
plt.title("Average cummulative sales across all restaurent ")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()


```
